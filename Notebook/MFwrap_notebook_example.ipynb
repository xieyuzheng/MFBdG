This uses a 6x6 repeated cell and a 3x3 momentum mesh (18x18 periodic dimensions).
Use `Boundary_Condition: "PBC"` (default); `"OBC"` prints a message and stops.
Set `Nkx=Nky=1` (the default) for a finite periodic 6x6 cell. `diag_threads` is the only thread input; allocation
is printed first. `pure_SC` allows both singlet and triplet bond pairing; use
`"pairing_sector": "singlet"` for a persistent singlet restriction.

The reported runtime is total wrapper time. After `%run`, `output` points to
the result file; the inspection cell below checks its convergence. A saved
state is not necessarily converged. See [the introduction](../README.md) and
[input guide](../Input/INPUT_GUIDE.md) for constraints, memory use, and examples.

````python
# Inspect the saved state, including when the wrapper skips a completed run.
import json
import numpy as np
from CPU.Plot import plot_iteration_state
from CPU.MFSolver import stop_reason

with np.load(output) as state:
    parameters = json.loads(state["parameters"].item())
    iterate = int(state["iterate"])
    err = float(state["err"])
    print(f"n = {state['filling']:.9f}, mu = {state['mu']:.8f}")
    print(f"iterate = {iterate}, err = {err:.6g}, tolerance = {parameters['tolerate']:g}")
    reason = stop_reason(state, parameters)
    print("Converged" if reason == "converged" else f"Unconverged ({reason})")
    if parameters.get("Plotting"):
        plot_iteration_state(parameters, iterate, state)
````
