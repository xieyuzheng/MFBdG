# Free fermions
Run both code cells with **Python (MFBdG)**, then release a slider to update the band structure, Fermi contour and DOS.

Energy is the bare band energy $\varepsilon$; dashed lines mark the **zero-temperature Fermi energy** $\mu$ at the selected filling (including both spins). The default k-grid is **1200 x 1200**; use a smaller `Nk` for faster exploration. Plots appear here without saving a file.

```python
%matplotlib inline
from pathlib import Path
import sys

import ipywidgets as widgets
from IPython.display import display

# Works when Jupyter starts in either MFBdG/ or free_fermion/.
project = Path.cwd()
if project.name == "free_fermion":
    project = project.parent
sys.path.insert(0, str(project))
from free_fermion.bands import number_solution, plot_bands
```

```python
# Edit the slider values, limits or steps here if needed.
controls = {
    "lattice": widgets.Dropdown(options=["Square", "Triangle"], value="Square", description="Lattice"),
    "t1": widgets.FloatSlider(value=-1, min=-3, max=3, step=0.05, description="t1", continuous_update=False),
    "t2": widgets.FloatSlider(value=0, min=-3, max=3, step=0.05, description="t2", continuous_update=False),
    "t3": widgets.FloatSlider(value=0, min=-3, max=3, step=0.05, description="t3", continuous_update=False),
    "n_filling": widgets.FloatSlider(value=0.9, min=0.01, max=1.99, step=0.01, description="n_filling", continuous_update=False),
    "Nk": widgets.IntSlider(value=1200, min=120, max=2400, step=120, description="Nk", continuous_update=False),
}


def update(lattice, t1, t2, t3, n_filling, Nk):
    hopping = (t1, t2, t3)
    mu, band = number_solution(lattice, hopping, n_filling, Nk)
    print(f"n = {n_filling:.10f}, T = 0, mu = {mu:.10f}")
    plot_bands(lattice, hopping, n_filling, mu, Nk, band)


output = widgets.interactive_output(update, controls)
display(widgets.VBox(list(controls.values())), output)
```
